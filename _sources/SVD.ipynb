````python
import numpy as np

# Matriks 3x3
A = np.array([
    [3, 1, 1],
    [-1, 3, 1],
    [1, 1, 2]
], dtype=float)

# SVD
U, S, VT = np.linalg.svd(A)

# Ubah singular values menjadi matriks diagonal
Sigma = np.diag(S)

print("U =")
print(U)

print("\nSigma =")
print(Sigma)

print("\nV^T =")
print(VT)

# Verifikasi A = U Σ V^T
A_rekonstruksi = U @ Sigma @ VT

print("\nU @ Sigma @ V^T =")
print(A_rekonstruksi)

print("\nA asli =")
print(A)
````

## Singular Value Decomposition

Dekomposisi Singular Value (SVD) menyatakan bahwa suatu matriks \(A\) dapat ditulis sebagai:

\[
A = U \Sigma V^T
\]

dengan:

- \(U\) = matriks ortogonal kiri
- \(\Sigma\) = matriks diagonal yang berisi singular value
- \(V^T\) = transpose dari matriks ortogonal kanan

---

## Langkah Perhitungan

1. Hitung dekomposisi SVD menggunakan:

\[
A = U \Sigma V^T
\]

2. Bentuk matriks diagonal \(\Sigma\) dari singular values.

3. Verifikasi hasil dengan menghitung:

\[
U \Sigma V^T
\]

4. Bandingkan hasil rekonstruksi dengan matriks asli \(A\).

---

## Verifikasi

Jika hasil rekonstruksi memenuhi:

\[
U \Sigma V^T = A
\]

maka dekomposisi SVD telah dilakukan dengan benar.

---

## Kesimpulan

Metode Singular Value Decomposition (SVD) berhasil memfaktorkan matriks \(A\) menjadi:

\[
A = U \Sigma V^T
\]

dan hasil perkalian kembali

\[
U \Sigma V^T
\]

menghasilkan matriks yang sama dengan matriks asli \(A\), sehingga dekomposisi terbukti benar.
