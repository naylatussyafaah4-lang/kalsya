# Dekomposisi QR dengan Metode Gram–Schmidt

Diberikan matriks

$$
A=
\begin{bmatrix}
1 & 2 & 0 & 1\\
2 & 1 & 1 & 0\\
0 & 1 & 2 & 1\\
1 & 0 & 1 & 2
\end{bmatrix}
$$

Kolom-kolom matriks:

$$
a_1=
\begin{bmatrix}
1\\2\\0\\1
\end{bmatrix},
\quad
a_2=
\begin{bmatrix}
2\\1\\1\\0
\end{bmatrix},
\quad
a_3=
\begin{bmatrix}
0\\1\\2\\1
\end{bmatrix},
\quad
a_4=
\begin{bmatrix}
1\\0\\1\\2
\end{bmatrix}
$$

---

# Langkah 1 — Menentukan $q_1$

Hitung norma:

$$
\|a_1\|
=
\sqrt{1^2+2^2+0^2+1^2}
=
\sqrt{6}
$$

Normalisasi:

$$
q_1
=
\frac{a_1}{\|a_1\|}
=
\frac{1}{\sqrt6}
\begin{bmatrix}
1\\2\\0\\1
\end{bmatrix}
$$

---

# Langkah 2 — Menentukan $q_2$

Hitung proyeksi $a_2$ terhadap $q_1$:

$$
a_2\cdot q_1
=
\frac{1}{\sqrt6}(2\cdot1+1\cdot2+1\cdot0+0\cdot1)
=
\frac{4}{\sqrt6}
$$

Cari residu:

$$
v_2
=
a_2-(a_2\cdot q_1)q_1
$$

$$
=
\begin{bmatrix}
2\\1\\1\\0
\end{bmatrix}
-
\frac{4}{6}
\begin{bmatrix}
1\\2\\0\\1
\end{bmatrix}
$$

$$
=
\begin{bmatrix}
2-\frac23\\
1-\frac43\\
1\\
0-\frac23
\end{bmatrix}
=
\begin{bmatrix}
\frac43\\
-\frac13\\
1\\
-\frac23
\end{bmatrix}
$$

Norma:

$$
\|v_2\|
=
\sqrt{
\left(\frac43\right)^2+
\left(-\frac13\right)^2+
1^2+
\left(-\frac23\right)^2
}
$$

$$
=
\sqrt{
\frac{16}{9}+\frac19+1+\frac49
}
=
\sqrt{\frac{30}{9}}
=
\sqrt{\frac{10}{3}}
$$

Normalisasi:

$$
q_2
=
\frac{v_2}{\|v_2\|}
=
\sqrt{\frac{3}{10}}
\begin{bmatrix}
\frac43\\
-\frac13\\
1\\
-\frac23
\end{bmatrix}
$$

atau setara dengan

$$
q_2
=
\frac{1}{\sqrt{30}}
\begin{bmatrix}
4\\-1\\3\\-2
\end{bmatrix}
$$

---

# Langkah 3 — Menentukan $q_3$

Gunakan:

$$
v_3
=
a_3
-(a_3\cdot q_1)q_1
-(a_3\cdot q_2)q_2
$$

Hitung:

$$
a_3\cdot q_1
=
\frac{1}{\sqrt6}(0+2+0+1)
=
\frac{3}{\sqrt6}
$$

Hitung:

$$
a_3\cdot q_2
=
\frac{1}{\sqrt{30}}
(0\cdot4+1(-1)+2(3)+1(-2))
=
\frac{3}{\sqrt{30}}
$$

Substitusi:

$$
v_3
=
a_3
-
\frac{3}{\sqrt6}q_1
-
\frac{3}{\sqrt{30}}q_2
$$

Diperoleh:

$$
v_3
=
\begin{bmatrix}
-1\\0\\2\\1
\end{bmatrix}
$$

Norma:

$$
\|v_3\|
=
\sqrt{(-1)^2+0^2+2^2+1^2}
=
\sqrt6
$$

Normalisasi:

$$
q_3
=
\frac{1}{\sqrt6}
\begin{bmatrix}
-1\\0\\2\\1
\end{bmatrix}
$$

---

# Langkah 4 — Menentukan $q_4$

Gunakan:

$$
v_4
=
a_4
-(a_4\cdot q_1)q_1
-(a_4\cdot q_2)q_2
-(a_4\cdot q_3)q_3
$$

Hitung:

$$
a_4\cdot q_1
=
\frac{3}{\sqrt6}
$$

$$
a_4\cdot q_2
=
-\frac{3}{\sqrt{30}}
$$

$$
a_4\cdot q_3
=
\frac{3}{\sqrt6}
$$

Substitusi menghasilkan:

$$
v_4
=
\begin{bmatrix}
0\\-1\\0\\1
\end{bmatrix}
$$

Norma:

$$
\|v_4\|
=
\sqrt2
$$

Normalisasi:

$$
q_4
=
\frac{1}{\sqrt2}
\begin{bmatrix}
0\\-1\\0\\1
\end{bmatrix}
$$

---

# Membentuk Matriks $Q$

$$
Q=
\begin{bmatrix}
\dfrac1{\sqrt6} & \dfrac4{\sqrt{30}} & -\dfrac1{\sqrt6} & 0\\[6pt]
\dfrac2{\sqrt6} & -\dfrac1{\sqrt{30}} & 0 & -\dfrac1{\sqrt2}\\[6pt]
0 & \dfrac3{\sqrt{30}} & \dfrac2{\sqrt6} & 0\\[6pt]
\dfrac1{\sqrt6} & -\dfrac2{\sqrt{30}} & \dfrac1{\sqrt6} & \dfrac1{\sqrt2}
\end{bmatrix}
$$

---

# Membentuk Matriks $R$

Gunakan:

$$
R=Q^TA
$$

Hasil:

$$
R=
\begin{bmatrix}
\sqrt6 & \dfrac4{\sqrt6} & \dfrac3{\sqrt6} & \dfrac3{\sqrt6}\\[8pt]
0 & \sqrt{\dfrac{10}{3}} & \dfrac3{\sqrt{30}} & -\dfrac3{\sqrt{30}}\\[8pt]
0 & 0 & \sqrt6 & \dfrac3{\sqrt6}\\[8pt]
0 & 0 & 0 & \sqrt2
\end{bmatrix}
$$

---

# Verifikasi

Kalikan:

$$
QR
$$

Maka diperoleh:

$$
QR=
\begin{bmatrix}
1 & 2 & 0 & 1\\
2 & 1 & 1 & 0\\
0 & 1 & 2 & 1\\
1 & 0 & 1 & 2
\end{bmatrix}
=A
$$

Karena

$$
QR=A
$$

maka dekomposisi QR yang diperoleh benar.
