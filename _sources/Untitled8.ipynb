```python
import matplotlib.pyplot as plt
import matplotlib.animation as animation
from IPython.display import HTML, display
import matplotlib as mpl

# Naikkan batas ukuran animasi
mpl.rcParams['animation.embed_limit'] = 100

# Titik
points = {
    "A": (2, 3), "B": (2, 4), "C": (3, 4), "D": (3, 3),
    "E": (2, -3), "F": (3, -3), "G": (2, -4), "H": (3, -4),
    "I": (2, 2), "J": (3, 2), "K": (2, 1), "L": (3, 1),
    "M": (2, -1), "N": (3, -1), "O": (2, -2), "P": (3, -2)
}

# Garis penghubung
garis = [
    ("B","C"),("A","D"),("I","J"),("K","L"),
    ("M","N"),("O","P"),("E","F"),("H","G"),
    ("B","A"),("A","I"),("I","K"),
    ("C","D"),("D","J"),("J","L"),
    ("M","O"),("O","E"),
    ("N","P"),("P","F")
]

fig, ax = plt.subplots(figsize=(6,8))

def update(frame):
    ax.clear()

    # Animasi mendekati lalu menjauhi sumbu X
    scale_y = abs(1 - frame / 25)

    posisi = {}

    for nama, (x, y) in points.items():
        y_baru = y * scale_y
        posisi[nama] = (x, y_baru)

        ax.scatter(x, y_baru, color="blue", s=80)
        ax.text(
            x + 0.05,
            y_baru + 0.05,
            nama,
            fontsize=9
        )

    for p1, p2 in garis:
        x1, y1 = posisi[p1]
        x2, y2 = posisi[p2]

        ax.plot(
            [x1, x2],
            [y1, y2],
            color="red",
            linewidth=2
        )

    ax.axhline(0, color="black", linewidth=1.5)
    ax.axvline(0, color="black", linewidth=1.5)

    ax.set_xlim(0, 5)
    ax.set_ylim(-5, 5)
    ax.grid(True)

    ax.set_title("Transformasi Mendekati Sumbu X")

ani = animation.FuncAnimation(
    fig,
    update,
    frames=50,
    interval=80,
    repeat=True
)

plt.close(fig)

display(HTML(ani.to_jshtml()))
```
