## A. Determinan Matriks 3×3

### Soal 1

Diketahui:

$$
A=
\begin{bmatrix}
2 & 1 & 3\\
0 & 4 & 5\\
1 & 2 & 1
\end{bmatrix}
$$

Rumus determinan matriks 3×3:

$$
\det(A)=a(ei-fh)-b(di-fg)+c(dh-eg)
$$

Maka:

$$
\det(A)=2(4\cdot1-5\cdot2)-1(0\cdot1-5\cdot1)+3(0\cdot2-4\cdot1)
$$

$$
\det(A)=2(4-10)-1(0-5)+3(0-4)
$$

$$
\det(A)=2(-6)-1(-5)+3(-4)
$$

$$
\det(A)=-12+5-12
$$

$$
\det(A)=-19
$$

Jadi:

$$
\boxed{\det(A)=-19}
$$

### Soal 3

Diketahui:

$$
C=
\begin{bmatrix}
1 & 2 & 3\\
2 & 4 & 6\\
1 & 1 & 1
\end{bmatrix}
$$

Perhatikan baris pertama dan baris kedua:

$$
R_2 = 2R_1
$$

Karena baris kedua merupakan kelipatan dari baris pertama, maka baris-baris pada matriks tidak bebas linear.

Akibatnya:

$$
\det(C)=0
$$

Jadi:

$$
\boxed{\det(C)=0}
$$

Karena determinannya sama dengan nol, maka matriks tersebut adalah matriks singular.

$$
\boxed{\text{Matriks C adalah matriks singular}}
$$

## B. Dekomposisi Matriks LU

### Soal 4

Diketahui:

$$
A=
\begin{bmatrix}
2 & 4 & 2\\
1 & 5 & 2\\
1 & 2 & 4
\end{bmatrix}
$$

Akan dicari:

$$
A=LU
$$

dengan \(L\) matriks segitiga bawah dan \(U\) matriks segitiga atas.

Matriks awal:

$$
\begin{bmatrix}
2 & 4 & 2\\
1 & 5 & 2\\
1 & 2 & 4
\end{bmatrix}
$$

Eliminasi baris kedua:

$$
m_{21}=\frac{1}{2}
$$

$$
R_2 \leftarrow R_2-\frac{1}{2}R_1
$$

$$
R_2=
\begin{bmatrix}
1 & 5 & 2
\end{bmatrix}
-
\frac{1}{2}
\begin{bmatrix}
2 & 4 & 2
\end{bmatrix}
$$

$$
R_2=
\begin{bmatrix}
0 & 3 & 1
\end{bmatrix}
$$

Eliminasi baris ketiga:

$$
m_{31}=\frac{1}{2}
$$

$$
R_3 \leftarrow R_3-\frac{1}{2}R_1
$$

$$
R_3=
\begin{bmatrix}
1 & 2 & 4
\end{bmatrix}
-
\frac{1}{2}
\begin{bmatrix}
2 & 4 & 2
\end{bmatrix}
$$

$$
R_3=
\begin{bmatrix}
0 & 0 & 3
\end{bmatrix}
$$

Maka diperoleh:

$$
U=
\begin{bmatrix}
2 & 4 & 2\\
0 & 3 & 1\\
0 & 0 & 3
\end{bmatrix}
$$

Nilai multiplier dimasukkan ke matriks \(L\):

$$
L=
\begin{bmatrix}
1 & 0 & 0\\
\frac{1}{2} & 1 & 0\\
\frac{1}{2} & 0 & 1
\end{bmatrix}
$$

Jadi hasil dekomposisi LU adalah:

$$
\boxed{
L=
\begin{bmatrix}
1 & 0 & 0\\
\frac{1}{2} & 1 & 0\\
\frac{1}{2} & 0 & 1
\end{bmatrix}
}
$$

$$
\boxed{
U=
\begin{bmatrix}
2 & 4 & 2\\
0 & 3 & 1\\
0 & 0 & 3
\end{bmatrix}
}
$$

Sehingga:

$$
\boxed{A=LU}
$$

$$
m_{21}=\frac{2}{1}=2

R_2=
\begin{bmatrix}

\end{bmatrix}
-

\end{bmatrix}

$$
R_2=
\begin{bmatrix}
0 & 1 & 1
\end{bmatrix}
$$

$$
m_{31}=\frac{4}{1}=4
$$

$$
R_3 \leftarrow R_3-4R_1
$$

$$
R_3=
\begin{bmatrix}
4 & 10 & 8
\end{bmatrix}
-
4
\begin{bmatrix}
1 & 2 & 1
\end{bmatrix}
$$

$$
R_3=
\begin{bmatrix}
0 & 2 & 4
\end{bmatrix}
$$

Selanjutnya eliminasi baris ketiga menggunakan baris kedua:

$$
m_{32}=\frac{2}{1}=2
$$

$$
R_3 \leftarrow R_3-2R_2
$$

$$
R_3=
\begin{bmatrix}
0 & 2 & 4
\end{bmatrix}
-
2
\begin{bmatrix}
0 & 1 & 1
\end{bmatrix}
$$

$$
R_3=
\begin{bmatrix}

\end{bmatrix}
$$

Maka diperoleh:

$$

\begin{bmatrix}
1 & 2 & 1\\
0 & 1 & 1\\
0 & 0 & 2
\end{bmatrix}

Nilai multiplier dimasukkan ke matriks \(L\):

$$
L=
\begin{bmatrix}
1 & 0 & 0\\

Jadi:

$$
\boxed{
L=
\begin{bmatrix}
1 & 0 & 0\\
2 & 1 & 0\\
4 & 2 & 1
\end{bmatrix}

$$

$$
\boxed{
U=
\begin{bmatrix}
1 & 2 & 1\\

$$

Sehingga:

$$
\boxed{B=LU}

## C. Invers Matriks 3×3

### Soal 7

Diketahui:

$$
A=
\begin{bmatrix}
1 & 2 & 1\\
0 & 1 & 1\\
2 & 3 & 4
\end{bmatrix}
$$

Akan dicari invers menggunakan metode Gauss-Jordan.

Bentuk matriks augmented:

$$
\left[
\begin{array}{ccc|ccc}
1 & 2 & 1 & 1 & 0 & 0\\
0 & 1 & 1 & 0 & 1 & 0\\
2 & 3 & 4 & 0 & 0 & 1
\end{array}
\right]
$$

Eliminasi elemen baris ketiga kolom pertama:

$$
R_3 \leftarrow R_3-2R_1
$$

$$
\left[
\begin{array}{ccc|ccc}
1 & 2 & 1 & 1 & 0 & 0\\
0 & 1 & 1 & 0 & 1 & 0\\
0 & -1 & 2 & -2 & 0 & 1
\end{array}
\right]
$$

Eliminasi elemen baris pertama dan ketiga kolom kedua:

$$
R_1 \leftarrow R_1-2R_2
$$

$$
R_3 \leftarrow R_3+R_2
$$

$$
\left[
\begin{array}{ccc|ccc}
1 & 0 & -1 & 1 & -2 & 0\\
0 & 1 & 1 & 0 & 1 & 0\\
0 & 0 & 3 & -2 & 1 & 1
\end{array}
\right]
$$

Ubah pivot baris ketiga menjadi 1:

$$
R_3 \leftarrow \frac{1}{3}R_3
$$

$$
\left[
\begin{array}{ccc|ccc}
1 & 0 & -1 & 1 & -2 & 0\\
0 & 1 & 1 & 0 & 1 & 0\\
0 & 0 & 1 & -\frac{2}{3} & \frac{1}{3} & \frac{1}{3}
\end{array}
\right]
$$

Eliminasi kolom ketiga:

$$
R_1 \leftarrow R_1+R_3
$$

$$
R_2 \leftarrow R_2-R_3
$$

$$
\left[
\begin{array}{ccc|ccc}
1 & 0 & 0 & \frac{1}{3} & -\frac{5}{3} & \frac{1}{3}\\
0 & 1 & 0 & \frac{2}{3} & \frac{2}{3} & -\frac{1}{3}\\
0 & 0 & 1 & -\frac{2}{3} & \frac{1}{3} & \frac{1}{3}
\end{array}
\right]
$$

Jadi:

$$
\boxed{
A^{-1}=
\begin{bmatrix}
\frac{1}{3} & -\frac{5}{3} & \frac{1}{3}\\
\frac{2}{3} & \frac{2}{3} & -\frac{1}{3}\\
-\frac{2}{3} & \frac{1}{3} & \frac{1}{3}
\end{bmatrix}
}
$$

\left[
