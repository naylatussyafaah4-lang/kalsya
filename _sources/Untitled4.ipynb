# Dekomposisi QR dengan Metode Gram-Schmidt

Diberikan matriks

$$
A=
\begin{bmatrix}
2 & 0 & 1 & 3\\
1 & 2 & 0 & 1\\
3 & 1 & 2 & 0\\
0 & 1 & 3 & 2
\end{bmatrix}
$$

Kolom-kolom matriks:

$$
a_1=
\begin{bmatrix}
2\\1\\3\\0
\end{bmatrix},
\quad
a_2=
\begin{bmatrix}
0\\2\\1\\1
\end{bmatrix},
\quad
a_3=
\begin{bmatrix}
1\\0\\2\\3
\end{bmatrix},
\quad
a_4=
\begin{bmatrix}
3\\1\\0\\2
\end{bmatrix}
$$

---

# Langkah 1: Vektor Pertama

Hitung norma $a_1$:

$$
\|a_1\|
=
\sqrt{2^2+1^2+3^2+0^2}
=
\sqrt{14}
$$

Normalisasi:

$$
q_1
=
\frac{a_1}{\|a_1\|}
=
\frac{1}{\sqrt{14}}
\begin{bmatrix}
2\\1\\3\\0
\end{bmatrix}
$$

---

# Langkah 2: Vektor Kedua

Hitung proyeksi:

$$
a_2 \cdot q_1
=
\frac{0(2)+2(1)+1(3)+1(0)}{\sqrt{14}}
=
\frac{5}{\sqrt{14}}
$$

Cari residu:

$$
v_2
=
a_2-(a_2\cdot q_1)q_1
$$

$$
v_2
=
\begin{bmatrix}
0\\2\\1\\1
\end{bmatrix}
-
\frac{5}{14}
\begin{bmatrix}
2\\1\\3\\0
\end{bmatrix}
=
\begin{bmatrix}
-\frac57\\[2mm]
\frac{23}{14}\\[2mm]
-\frac1{14}\\[2mm]
1
\end{bmatrix}
$$

Norma:

$$
\|v_2\|
=
\sqrt{\left(-\frac57\right)^2+\left(\frac{23}{14}\right)^2+\left(-\frac1{14}\right)^2+1^2}
=
\sqrt{\frac52}
$$

Normalisasi:

$$
q_2
=
\frac{v_2}{\|v_2\|}
$$

$$
q_2
=
\frac{1}{\sqrt{70}}
\begin{bmatrix}
-2\\[2mm]
\frac{23}{5}\\[2mm]
-\frac15\\[2mm]
\frac{14}{5}
\end{bmatrix}
$$

---

# Langkah 3: Vektor Ketiga

Gunakan:

$$
v_3
=
a_3-(a_3\cdot q_1)q_1-(a_3\cdot q_2)q_2
$$

Hitung:

$$
a_3\cdot q_1
=
\frac{1(2)+0(1)+2(3)+3(0)}{\sqrt{14}}
=
\frac8{\sqrt{14}}
$$

dan

$$
a_3\cdot q_2
=
\frac3{\sqrt{70}}
$$

Maka:

$$
v_3
=
\begin{bmatrix}
-\frac13\\[2mm]
-\frac13\\[2mm]
\frac13\\[2mm]
\frac43
\end{bmatrix}
$$

Norma:

$$
\|v_3\|
=
\sqrt2
$$

Normalisasi:

$$
q_3
=
\frac{v_3}{\|v_3\|}
=
\frac{1}{3\sqrt2}
\begin{bmatrix}
-1\\-1\\1\\4
\end{bmatrix}
$$

---

# Langkah 4: Vektor Keempat

Gunakan:

$$
v_4
=
a_4-(a_4\cdot q_1)q_1-(a_4\cdot q_2)q_2-(a_4\cdot q_3)q_3
$$

Hasil:

$$
v_4
=
\begin{bmatrix}
1\\-1\\1\\-1
\end{bmatrix}
$$

Norma:

$$
\|v_4\|
=
2
$$

Normalisasi:

$$
q_4
=
\frac12
\begin{bmatrix}
1\\-1\\1\\-1
\end{bmatrix}
$$

---

# Membentuk Matriks $Q$

$$
Q=
\begin{bmatrix}
\frac{2}{\sqrt{14}} & -\frac{2}{\sqrt{70}} & -\frac{1}{3\sqrt2} & \frac12\\
\frac{1}{\sqrt{14}} & \frac{23}{5\sqrt{70}} & -\frac{1}{3\sqrt2} & -\frac12\\
\frac{3}{\sqrt{14}} & -\frac{1}{5\sqrt{70}} & \frac{1}{3\sqrt2} & \frac12\\
0 & \frac{14}{5\sqrt{70}} & \frac{4}{3\sqrt2} & -\frac12
\end{bmatrix}
$$

---

# Membentuk Matriks $R$

$$
R=Q^TA
$$

$$
R=
\begin{bmatrix}
\sqrt{14} & \frac5{\sqrt{14}} & \frac8{\sqrt{14}} & \frac7{\sqrt{14}}\\
0 & \sqrt{\frac52} & \frac3{\sqrt{70}} & \frac{16}{\sqrt{70}}\\
0 & 0 & \sqrt2 & \frac5{\sqrt2}\\
0 & 0 & 0 & 2
\end{bmatrix}
$$

---

# Hasil Akhir

$$
A = QR
$$

dengan

$$
Q^TQ = I
$$

dan $R$ matriks segitiga atas.
